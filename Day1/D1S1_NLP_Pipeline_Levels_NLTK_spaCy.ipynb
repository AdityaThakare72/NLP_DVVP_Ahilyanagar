## 1. Ambiguity

Same word, different meanings. WordNet stores each meaning as a synset.

```python
from nltk.corpus import wordnet as wn

# print the first few senses of 'bank'
for s in wn.synsets("bank")[:4]:
    print(s.name(), "->", s.definition())
```

```python
# syntactic ambiguity: same words, two structures
sentence = "I saw a man with a telescope"
readings = {
    "reading 1": "I used a telescope to see the man",
    "reading 2": "I saw a man who was holding a telescope",
}
for name, meaning in readings.items():
    print(name, ":", meaning)
```

## 2. Levels of language

spaCy shows the syntax level directly: a tag and a dependency label for each word.

```python
doc = nlp("Can you pass the salt?")
for t in doc:
    print(t.text, t.pos_, t.dep_)

# syntax tells us this is a question
# pragmatics (it is a polite request) is not in the output at all
```

## 3. The pipeline, stage by stage

We go from raw text to cleaned tokens and print the result after every stage.

```python
stopwords.words("french")
```

```python
from nltk.corpus import stopwords

raw = "The Cats were chasing mice! Dr. Rao watched them."

sentences = nltk.sent_tokenize(raw)
print("sentences:", sentences)

tokens = nltk.word_tokenize(raw)
print("tokens:", tokens)

lower = [t.lower() for t in tokens if t.isalpha()]
print("lowercase words:", lower)

stop = set(stopwords.words("english"))
content = [t for t in lower if t not in stop]
print("without stopwords:", content)

print("pos tags:", nltk.pos_tag(content))
```

## 4. NLTK and spaCy on the same text

```python
text = "Dr. Rao isn't coming to the U.S. office."
print("nltk :", nltk.word_tokenize(text))
print("spacy:", [t.text for t in nlp(text)])
```

```python
# messier text shows the differences better
messy = "mail me at rao@example.com #nlp costs $5.99 :)"
print("nltk :", nltk.word_tokenize(messy))
print("spacy:", [t.text for t in nlp(messy)])
```

```python
# spacy gives pos and lemma in the same call
text = "The cats were chasing mice in the garden."
for t in nlp(text):
    print(f"{t.text:10} {t.pos_:6} {t.lemma_}")

# nltk needs a separate call for tags
print(nltk.pos_tag(nltk.word_tokenize(text)))
```

## 5. A small reusable function

We wrap the stages in one function. We will extend this on the next sessions.

```python
def analyze(text):
    # returns every stage so we can inspect them
    tokens = nltk.word_tokenize(text)
    words = [t.lower() for t in tokens if t.isalpha()]
    stop = set(stopwords.words("english"))
    content = [w for w in words if w not in stop]
    return {"tokens": tokens, "words": words, "content": content}

for sentence in ["Students are learning NLP at the workshop.", 
                 "I did not like the movie."]:
    print(analyze(sentence))
    print()

# look at the second one: 'not' is gone after stopword removal
# the meaning flipped, we will come back to this on the next session
```
