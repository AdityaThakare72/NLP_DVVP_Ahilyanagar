# Sentiment-Analysis

```python
model = pipeline(task= 'sentiment-analysis',
                 model = 'distilbert-base-uncased-finetuned-sst-2-english')
```

```python
model(['this was a great movie',
       'the acting was bad done by the actors',
       'the movie was so good that the plot went on a vaccation midway'])
```

# Zero shot classification

```python
# classify test into labels the model was never explicitly trained on
```

```python
cat = pipeline(task = "zero-shot-classification")
```

```python
cat(['he was doing biological experiment',
     'the game was so good and the graphics were nice',
     'nano-tech journal',
     'this act was passed in the recent bill',
     'she is calculating the cricket match projected score'],
    candidate_labels = ['science', 'gaming', 'law', 'economics', 'maths', 'sports'])
```

# text-generation
