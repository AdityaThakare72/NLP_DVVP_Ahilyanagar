```python
# sitting (a) -> kitten  (b)
```

```python
# since n is of 7 letters and m of 6
# need a grid 8 rows and 7 columns  (1 extra because of null)
```

```python
source = "sitting"
target = "kitten"
```

```python
n = len(source)
m = len(target)
n, m
```

```python
import numpy as np
```

```python
D = np.zeros((n + 1, m + 1), dtype=int)
D
```

```python
D[:, 0] = np.arange(n + 1)  # left column: 0,1,2,... (deletes)

D[0, :] = np.arange(m + 1)      # top row: 0,1,2,... (inserts)
```

```python
D
```

```python
for i in range(1, n + 1):
        for j in range(1, m + 1):
            sub = 0 if source[i-1] == target[j-1] else 1
            D[i, j] = min(D[i-1, j] + 1,        # above -> delete
                          D[i, j-1] + 1,        # left  -> insert
                          D[i-1, j-1] + sub)    # diagonal -> replace / match
```

```python
""" 
Trace one cell from your sheet. Take row t (the 3rd letter of sitting), column t (the 3rd letter of kitten). You wrote 1 there.

Above is 2, so delete gives 2+1 = 3.
Left is 2, so insert gives 2+1 = 3.
Diagonal is 1 and t == t, so sub = 0, giving 1+0 = 1.
The min is 1.
"""
D
```
